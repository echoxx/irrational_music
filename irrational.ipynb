```python
def calculate_frequencies(base_frequency, num_notes):
    frequencies = []
    ratio = 2 ** (1 / num_notes)
    
    for i in range(num_notes):
        frequency = base_frequency * (ratio ** i)
        frequencies.append(frequency)
        
    return frequencies
```

```python
def generate_audio(frequencies, duration=1, sample_rate=44100):
    t = np.linspace(0, duration, duration * sample_rate, False)
    audio = np.array([], dtype=np.int16)

    for frequency in frequencies:
        sine_wave = 0.5 * np.sin(2 * np.pi * frequency * t) # calculates y value
        sine_wave = (sine_wave * 32767).astype(np.int16) # convert to 16-bit signed int
        audio = np.concatenate((audio, sine_wave))

    return audio
```

```python
import numpy as np
import sounddevice as sd
from mpmath import mp

# Constants
sample_rate = 44100  # Sample rate in HZ
duration = 0.05  # Duration of each note in seconds
fade_time = 0.025  # Fade time in seconds
base_frequency = 440  # Base frequency in HZ (A4)
pi_digits = 10  # Number of digits of pi to consider

# Change precision and calculate pi
mp.dps = pi_digits + 1  # Set precision
pi = str(mp.pi)

# Calculate the frequencies
frequencies = [base_frequency * (2 ** (i/10)) for i in range(10)]

# Create a map of digit to frequency
digit_to_frequency = {str(i): f for i, f in enumerate(frequencies)}

# Map the digits of pi to frequencies
notes = [digit_to_frequency[d] for d in pi if d.isdigit()]

# Generate and play the sounds
for note in notes:
    # Generate the samples for this note
    samples = (np.sin(2 * np.pi * np.arange(sample_rate * duration) * note / sample_rate)).astype(np.float32)

    # Create the fade in and out
    fade_samples = int(fade_time * sample_rate)
    fade_in = np.linspace(0, 1, fade_samples)
    fade_out = np.linspace(1, 0, fade_samples)

    # Apply the fade in and out
    samples[:fade_samples] *= fade_in
    samples[-fade_samples:] *= fade_out

    # Play the note
    sd.play(samples, sample_rate)

    # Use sleep to prevent the note from stopping prematurely
    sd.sleep(int(duration * 1000))

# Stop the sound device
sd.stop()
```
